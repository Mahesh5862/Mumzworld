```python
%%writefile app.py
import streamlit as st

st.set_page_config(page_title="Mumz AI", layout="wide")

# 🎨 PREMIUM STYLE
st.markdown("""
<style>
.main {
    background: linear-gradient(to right, #fdfbfb, #ebedee);
}
.title {
    font-size: 40px;
    font-weight: bold;
    color: #ff4b6e;
}
.subtitle {
    color: gray;
    font-size: 18px;
}
.card {
    padding:20px;
    border-radius:20px;
    background:white;
    box-shadow:0px 6px 15px rgba(0,0,0,0.1);
}
</style>
""", unsafe_allow_html=True)

# 🍼 HEADER
col1, col2 = st.columns([1,5])
with col1:
    st.image("https://cdn-icons-png.flaticon.com/512/3771/3771544.png", width=90)
with col2:
    st.markdown('<div class="title">Mumz AI</div>', unsafe_allow_html=True)
    st.markdown('<div class="subtitle">Smart Baby Care Assistant</div>', unsafe_allow_html=True)

st.markdown("---")

# 🌍 LANGUAGE
lang = st.selectbox("🌍 Language / اللغة", ["English", "Arabic"])

# 📊 SIDEBAR (PREMIUM FEATURE)
with st.sidebar:
    st.header("⚙️ Preferences")
    age = st.selectbox("👶 Baby Age", ["Newborn", "6 months", "1 year", "2 years"])
    budget = st.selectbox("💰 Budget", ["Low", "Medium", "High"])

# 💬 INPUT
user_input = st.text_input(
    "Describe your need (e.g. stroller for travel)"
    if lang=="English" else
    "اكتبي ما تحتاجينه"
)

# 🧠 SMART LOGIC
def generate_response(text, age, budget):
    text = text.lower()

    if "stroller" in text:
        product = "Premium Travel Stroller"
        reason = f"Perfect for {age}, lightweight and safe"
    elif "toy" in text:
        product = "Educational Baby Toys"
        reason = f"Best for {age}, improves brain development"
    elif "book" in text:
        product = "Interactive Learning Books"
        reason = f"Great for {age}, builds early skills"
    elif "feeding" in text:
        product = "Smart Feeding Kit"
        reason = f"Helpful for {age}, safe feeding experience"
    else:
        product = "Recommended Baby Product"
        reason = f"Suitable for {age} within {budget} budget"

    tip = f"Always choose safe, certified products within {budget} budget"

    return product, reason, tip

# 🚀 BUTTON
if st.button("✨ Get Recommendation" if lang=="English" else "✨ احصلي على توصية"):

    product, reason, tip = generate_response(user_input, age, budget)

    # 🎯 OUTPUT CARD
    st.markdown('<div class="card">', unsafe_allow_html=True)

    st.markdown("## 🎯 Recommended Product" if lang=="English" else "## 🎯 المنتج المقترح")
    st.success(f"🛍️ {product}")

    st.markdown("### 💡 Why this?" if lang=="English" else "### 💡 لماذا هذا؟")
    st.write(reason)

    st.markdown("### ⚠️ Expert Tip" if lang=="English" else "### ⚠️ نصيحة")
    st.info(tip)

    st.markdown('</div>', unsafe_allow_html=True)

    st.markdown("### ⭐ Rating")
    st.write("⭐⭐⭐⭐☆ (4.5/5)")

    st.markdown("---")
    st.write("🤖 Mumz AI | Smart • Safe • Trusted")
```
